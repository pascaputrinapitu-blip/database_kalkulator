```
import pandas as pd
import numpy as np

from google.colab import files
uploaded = files.upload()

# Sesuaikan nama file jika berbeda
df = pd.read_csv("dataset_bonus_karyawan_180.csv")

print("Ukuran data:", df.shape)
display(df.head())
df.info()
```

```
print("Jumlah missing value:")
print(df.isna().sum())

print("\\nDistribusi target:")
print(df["layak_bonus"].value_counts())
print("\\nProporsi target:")
print(df["layak_bonus"].value_counts(normalize=True) * 100)

print("\\nNilai unik kolom kategorikal:")
for col in ["departemen", "status_karyawan"]:
    print(col, df[col].unique())
```

```
# Standarisasi nilai kosong
df = df.replace(["?", "", " "], np.nan)

numeric_cols = [
    "masa_kerja_tahun", "skor_kinerja",
    "persentase_kehadiran", "skor_disiplin",
    "jumlah_pelatihan"
]

#Konversi koma desimal menjadi titik, lalu ubah menjadi numerik
for col in numeric_cols:
    df[col] = (df[col].astype(str)
               .str.replace(",", ".", regex=False))
    df[col] = pd.to_numeric(df[col], errors="coerce")

categorical_cols = ["departemen", "status_karyawan"]
for col in categorical_cols:
    df[col] = df[col].replace("nan", np.nan)

# Pisahkan fitur dan target
X = df[numeric_cols + categorical_cols]
y = pd.to_numeric(df["layak_bonus"], errors="coerce")

# Pastikan baris dengan target kosong dikeluarkan
valid = y.notna()
X = X.loc[valid].copy()
y = y.loc[valid].astype(int)

print("Ukuran X:", X.shape)
print("Distribusi y:")
print(y.value_counts())
```

```
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Distribusi train:")
print(y_train.value_counts(normalize=True))
print("\\nDistribusi test:")
print(y_test.value_counts(normalize=True))
```

```
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Contoh: pilih fitur numerik saja
X_train_num = X_train[numeric_cols].copy()
X_test_num = X_test[numeric_cols].copy()

# 1. Pelajari median dari data training, lalu isi nilai kosong
imputer_num = SimpleImputer(strategy="median")
X_train_num_imp = imputer_num.fit_transform(X_train_num)
X_test_num_imp = imputer_num.transform(X_test_num)

# 2. Pelajari skala dari data training
scaler = StandardScaler()
X_train_num_scaled = scaler.fit_transform(X_train_num_imp)
X_test_num_scaled = scaler.transform(X_test_num_imp)

print("Training:", X_train_num_scaled.shape)
print("Testing:", X_test_num_scaled.shape)
```

```
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

X_train_cat = X_train[categorical_cols].copy()
X_test_cat = X_test[categorical_cols].copy()

imputer_cat = SimpleImputer(strategy="most_frequent")
X_train_cat_imp = imputer_cat.fit_transform(X_train_cat)
X_test_cat_imp = imputer_cat.transform(X_test_cat)

encoder = OneHotEncoder(handle_unknown="ignore")
X_train_cat_enc = encoder.fit_transform(X_train_cat_imp)
X_test_cat_enc = encoder.transform(X_test_cat_imp)

print("Training kategori:", X_train_cat_enc.shape)
print("Testing kategori:", X_test_cat_enc.shape)
```

```
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_cols),
    ("cat", categorical_transformer, categorical_cols)
])

baseline_model = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])

balanced_model = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000, class_weight="balanced"
    ))
])
```

```
from sklearn.metrics import (
    confusion_matrix, classification_report,
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

models = {
    "Baseline": baseline_model,
    "Class Weight Balanced": balanced_model
}

hasil = []

for nama, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    proba = model.predict_proba(X_test)[:, 1]

    hasil.append({
        "Model": nama,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred, zero_division=0),
        "F1-Score": f1_score(y_test, pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, proba)
    })

    print("\\nMODEL:", nama)
    print("Confusion Matrix:\\n", confusion_matrix(y_test, pred))
    print(classification_report(y_test, pred, zero_division=0))

hasil_df = pd.DataFrame(hasil)
display(hasil_df.round(3))
```

```
import matplotlib.pyplot as plt

plot_df = hasil_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1-Score", "ROC-AUC"]
]
plot_df.plot(kind="bar", figsize=(11, 5), ylim=(0, 1))
plt.title("Perbandingan Metrik Evaluasi")
plt.ylabel("Skor")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.25)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```
from sklearn.model_selection import KFold, StratifiedKFold, cross_validate

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

cv_schemes = {
    "KFold": KFold(n_splits=5, shuffle=True, random_state=42),
    "StratifiedKFold": StratifiedKFold(
        n_splits=5, shuffle=True, random_state=42
    )
}

cv_results = []

for nama_cv, cv in cv_schemes.items():
    scores = cross_validate(
        balanced_model, X_train, y_train,
        cv=cv, scoring=scoring,
        return_train_score=False
    )

    cv_results.append({
        "Skema": nama_cv,
        "Accuracy mean": scores["test_accuracy"].mean(),
        "Precision mean": scores["test_precision"].mean(),
        "Recall mean": scores["test_recall"].mean(),
        "F1 mean": scores["test_f1"].mean(),
        "F1 std": scores["test_f1"].std(),
        "ROC-AUC mean": scores["test_roc_auc"].mean()
    })

cv_df = pd.DataFrame(cv_results)
display(cv_df.round(3))
```

```
from sklearn.model_selection import KFold, StratifiedKFold

def cek_fold(nama, splitter, X_data, y_data):
    print("\\n", nama)
    for fold, (train_idx, val_idx) in enumerate(
        splitter.split(X_data, y_data), start=1
    ):
        proporsi = y_data.iloc[val_idx].value_counts(normalize=True)
        print(
            f"Fold {fold}:",
            "jumlah validasi =", len(val_idx),
            "| proporsi kelas =",
            proporsi.round(3).to_dict()
        )

cek_fold(
    "KFold",
    KFold(n_splits=5, shuffle=True, random_state=42),
    X_train, y_train
)

cek_fold(
    "StratifiedKFold",
    StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    X_train, y_train
)
```

5. Apakah model dengan F1-score tertinggi otomatis menjadi model terbaik untuk HR? Jelaskan berdasarkan konsekuensi FP dan FN.

Tidak selalu. F1-score tertinggi tidak otomatis menjadikan model terbaik untuk HR karena pemilihan model harus mempertimbangkan dampak kesalahan prediksi.

Ada dua jenis kesalahan yang perlu diperhatikan:
- False Positive (FP): Karyawan yang sebenarnya tidak layak bonus tetapi diprediksi layak bonus. Hal ini dapat menyebabkan perusahaan memberikan bonus kepada karyawan yang tidak memenuhi kriteria.
- False Negative (FN): Karyawan yang sebenarnya layak bonus tetapi diprediksi tidak layak bonus. Hal ini dapat merugikan karyawan yang seharusnya mendapatkan bonus.
Dalam kasus ini, jika HR lebih mengutamakan agar karyawan yang layak bonus tidak terlewat, maka recall perlu diprioritaskan.

Model Class Weight Balanced memiliki F1-score sebesar 0,667, lebih tinggi dibandingkan Baseline sebesar 0,333. Model ini juga memiliki False Negative lebih sedikit, yaitu 1 dibandingkan 3 pada Baseline.

Namun, model tersebut memiliki False Positive sebanyak 2, sedangkan Baseline hanya 1.

Jadi, HR perlu mempertimbangkan keseimbangan antara kesalahan memberikan bonus kepada karyawan yang tidak layak dan kesalahan melewatkan karyawan yang layak menerima bonus.

6. Jelaskan bagaimana data leakage dapat terjadi jika SMOTE dilakukan sebelum cross-validation.

Data leakage dapat terjadi ketika SMOTE dilakukan sebelum proses cross-validation karena data sintetis dibuat menggunakan informasi dari seluruh dataset sebelum data dibagi menjadi beberapa fold.

SMOTE bekerja dengan membuat data baru berdasarkan kemiripan atau hubungan antar data pada kelas minoritas.
